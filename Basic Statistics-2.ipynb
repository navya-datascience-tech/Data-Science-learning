Problem Statement : Hospital Patient Data Analysis

```python
import pandas as pd

patient_df = pd.read_csv("Patient_Data.csv")
billing_df = pd.read_csv("Billing_Data.csv")

patient_df.info()
```

```python
billing_columns = patient_df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
print(billing_columns.head())
```

```python
patient_df = patient_df.drop(columns=['ReceptionistID', 'CheckInTime'], errors='ignore')
patient_df 
```

```python
dept_bill = patient_df.groupby('Department')['BillAmount'].sum()

print(dept_bill)
```

```python
patient_df = patient_df.drop_duplicates(subset='PatientID')
patient_df
```

```python
mean_bill = patient_df['BillAmount'].mean()

patient_df['BillAmount'] = patient_df['BillAmount'].fillna(mean_bill)
patient_df
```

```python
merged_df = pd.merge(patient_df, billing_df, on='PatientID', how='inner')

print(merged_df.head())
```

```python
new_patients = pd.DataFrame({
    'PatientID': [201,202],
    'Department': ['Cardiology','Neurology'],
    'Doctor': ['Dr.A','Dr.B'],
    'BillAmount': [5000,7000]
})

patient_df = pd.concat([patient_df, new_patients], axis=0, ignore_index=True)
patient_df 
```

```python
new_columns = pd.DataFrame({
    'InsuranceCovered': ['Yes','No','Yes'],
    'FinalAmount': [4000,6500,5500]
})

patient_df = pd.concat([patient_df, new_columns], axis=1)
patient_df 
```

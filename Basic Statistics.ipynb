```python
num_cols = df.select_dtypes(include=['int64', 'float64'])
```

```python
print("Mean:\n", num_cols.mean())
```

```python
print("Median:\n", num_cols.median())
```

```python
print("Mode:\n", num_cols.mode().iloc[0])
```

```python
print("Standard Deviation:\n", num_cols.std())
```

### Data Visualization

```python
import matplotlib.pyplot as plt
num_cols.hist(figsize=(10,8),color="pink")
plt.show()
```

```python
import seaborn as sns

for col in num_cols:
    sns.boxplot(x=df[col])
    plt.show()
```

```python
cat_cols = df.select_dtypes(include=['object'])

for col in cat_cols:
    df[col].value_counts().plot(kind='bar')
    plt.title(col)
    plt.show()
```

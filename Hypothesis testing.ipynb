# ESTIMATION AND CONFIDENCE INTERVALS

```python
import numpy as np
import scipy.stats as stats

# Given data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07,
        1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

# Sample size
n = len(data)

# Sample mean
mean = np.mean(data)

# Sample standard deviation
std_sample = np.std(data, ddof=1)

# Confidence level
confidence = 0.99
alpha = 1 - confidence

# t critical value
t_value = stats.t.ppf(1 - alpha/2, df=n-1)

# Margin of error (t distribution)
margin_error_t = t_value * (std_sample / np.sqrt(n))

# Confidence interval using sample standard deviation
ci_t = (mean - margin_error_t, mean + margin_error_t)

print("Sample Mean:", mean)
print("Sample Standard Deviation:", std_sample)
print("99% Confidence Interval using t-distribution:", ci_t)
```

```python
# Known population standard deviation
sigma = 0.2

# Z critical value
z_value = stats.norm.ppf(1 - alpha/2)

# Margin of error (z distribution)
margin_error_z = z_value * (sigma / np.sqrt(n))

# Confidence interval using known population standard deviation
ci_z = (mean - margin_error_z, mean + margin_error_z)

print("99% Confidence Interval using z-distribution:", ci_z)
```

# HYPOTHESIS TESTING

```python
import numpy as np

# Given values
x_bar = 3050        # sample mean
mu = 1000 + 5*600   # theoretical mean
sigma = 5 * 25      # standard deviation
n = 25              # sample size

# Calculate Z statistic
z = (x_bar - mu) / (sigma / np.sqrt(n))

print("Theoretical Mean (μ):", mu)
print("Standard Deviation (σ):", sigma)
print("Test Statistic (Z):", z)
```

```python
from scipy.stats import norm

alpha = 0.05

# critical value
critical_value = norm.ppf(1 - alpha)

print("Critical Value:", critical_value)
```

```python
import numpy as np
from scipy.stats import norm

# Given values
x_bar = 3050
mu = 1000 + 5*600
sigma = 5 * 25
n = 25
alpha = 0.05

# Calculate Z statistic
z = (x_bar - mu) / (sigma / np.sqrt(n))

# Critical value
critical_value = norm.ppf(1 - alpha)

print("Z Statistic:", z)
print("Critical Value:", critical_value)

# Decision
if z > critical_value:
    print("Reject the Null Hypothesis (H0)")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
```

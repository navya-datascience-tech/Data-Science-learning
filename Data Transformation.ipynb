# DATA PREPROCESSING AND FEATURE ENGINEERING IN MACHINE LEARNING

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("adult_with_headers (1).csv")

# Display first rows
df.head()

# Dataset information
df.info()

# Summary statistics
df.describe()

# Check missing values
df.isnull().sum()
```

```python
### a.Standard Scaling
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Select numerical columns
num_cols = df.select_dtypes(include=['int64','float64']).columns

# Apply Standard Scaling
scaler = StandardScaler()
df_standard_scaled = df.copy()

df_standard_scaled[num_cols] = scaler.fit_transform(df_standard_scaled[num_cols])

df_standard_scaled.head()
```

```python
###  b. Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler

# Apply Min-Max Scaling
scaler = MinMaxScaler()

df_minmax_scaled = df.copy()

df_minmax_scaled[num_cols] = scaler.fit_transform(df_minmax_scaled[num_cols])

df_minmax_scaled.head()
```

# 2. Encoding Techniques:

# 3. Feature Engineering

```python

# Remove extra spaces in column names
df.columns = df.columns.str.strip()

# Replace spaces with underscore (optional but recommended)
df.columns = df.columns.str.replace(" ", "_")

# Check column names
print(df.columns)

# Feature 1: Work Intensity
df["work_intensity"] = df["hours_per_week"] / df["age"]

# Feature 2: Capital Gain Ratio
df["capital_ratio"] = df["capital_gain"] / (df["capital_loss"] + 1)

# View new features
df[["work_intensity","capital_ratio"]].head()
```

```python
import numpy as np

# Apply log transformation
df["capital_gain_log"] = np.log1p(df["capital_gain"])

# View transformed feature
df[["capital_gain", "capital_gain_log"]].head()
```

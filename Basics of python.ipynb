Exercise 1: Prime Numbers

```python
### checking prime num or not
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is NOT a Prime Number")
            break
    else:
        print(num, "is a Prime Number")
else:
    print(num, "is NOT a Prime Number")
```

Exercise 2: Product of Random Numbers

```python
#### Product of Random Numbers
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print("Number 1:", num1)
print("Number 2:", num2)

answer = int(input("Enter the product: "))
if answer == num1 * num2:
    print("Correct Answer!")
else:
    print("Wrong Answer")
    print("Correct product is:", num1 * num2)
```

Exercise 3: Squares of Even/Odd Numbers

```python
### Squares of Even Numbers from 100 to 200

for i in range(100, 201):
    if i % 2 == 0:  
        print("Number:", i, "Square:", i*i)
```

Exercise 4:Hospital Billing System

```python
### hospital charges
room_charge_per_day = 2000
days = 4
doctor_fee = 1500
lab_test_cost = 300
number_of_tests = 3
medicine_charges = 2400

# Calculations
room_total = room_charge_per_day * days
lab_total = lab_test_cost * number_of_tests

total_bill = room_total + doctor_fee + lab_total + medicine_charges

print("Total before discount:", total_bill)

if total_bill > 10000:
    discount = total_bill * 0.10
    final_bill = total_bill - discount
    print("Discount Applied:", discount)
else:
    final_bill = total_bill

print("Final Bill:", final_bill)
```
